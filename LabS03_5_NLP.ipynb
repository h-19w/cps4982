```python
text3 = "automate, automates, automatic, automation "
annotations = explain_document_pipeline.annotate(text3)
print(annotations['lemmas'])
print(annotations['stems'])
```

```python
import re
def removeSomePunctuation(text):
    return re.sub(r"[^a-z0-9\s]",'', text.lower().strip()) # only keep characters: a-z, 0-9 , and whitespace in a line
```

```python
# remove punctuations first
text4 = "automate, automates, automatic, automation "
nopunct_text = removeSomePunctuation(text4)
annotations = explain_document_pipeline.annotate(nopunct_text)
print(annotations['lemmas'])
print(annotations['stems'])
```
