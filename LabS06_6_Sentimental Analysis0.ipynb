```python
# pip install pyprind -i https://pypi.tuna.tsinghua.edu.cn/simple
import pyprind
import pandas as pd
import os
import numpy as np
```

## Dataset Preparation

```python
# read data from folders and save it to a csv file
# you may need to install pyprind first
# pip install pyprind

#there are 12500 neg files and 12500 pos files in train directory
#there are 12500 neg files and 12500 pos files in test directory
#total 12500x4 = 50000 files

pbar=pyprind.ProgBar(50000)
labels={'pos':1,'neg':0}
df=pd.DataFrame()
for s in ('test','train'):
    for l in ('pos','neg'):
        path='E:/aclimdb/%s/%s'% (s,l)
        for file in os.listdir(path):
            with open(os.path.join(path,file),'r', encoding='utf-8') as infile:
                txt=infile.read()
            df=df.append([[txt,labels[l]]],ignore_index=True)
            pbar.update()
df.columns=['review','sentiment']
```

```python
np.random.seed(0)
df=df.reindex(np.random.permutation(df.index))
df.to_csv('./data/movie_data.csv',index=False)
```

```python
idx = int(df.shape[0]*0.8)
train = df[:idx]
test = df[idx:]
train.to_csv('./data/movie_train.csv',index=False)
test.to_csv('./data/movie_test.csv',index=False)
print(train.shape, test.shape)
```
