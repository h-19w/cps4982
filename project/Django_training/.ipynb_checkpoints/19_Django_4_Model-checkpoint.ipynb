## 2. Models

A model is the single, definitive source of information about your data. It contains the essential fields and behaviors of the data you’re storing. Generally, each model maps to a single database table.

The basics:

- Each model is a Python class that subclasses ``django.db.models.Model``.
- Each attribute of the model represents a database field.
- With all of this, Django gives you an automatically-generated database-access API; see [Making queries](https://docs.djangoproject.com/en/3.0/topics/db/queries/).


### 2.1 Create Models
In Django, a model is a **class** that represents table or collection in our DB, and where every attribute of the class is a field of the table or collection. Models are defined in the ``wkunews/models.py``. Following is a User model created as an example:

```python
from django.db import models

# Create your models here.
class user (models.Model):
    username = models.CharField(max_length=25)
    password  = models.CharField(max_length=30)
    year = models.IntegerField()
    
    
class wkunews (models.Model):
    title = models.CharField(max_length=30)
    content = models.TextField()
    
```

- Here, the user model is represented by a class that subclasses django.db.models.Model. The model has a number of class variables, each of which represents a database field.

- Each field is represented by an instance of a Field class – e.g., CharField for character fields and IntegerField for integers. This tells Django what type of data each field holds.

- The name of each Field instance (e.g. username, password, year) is the field’s name, in machine-friendly format. You’ll use this value in your Python code, and your database will use it as the column name.

### 2.2 Activating models

That small bit of model code gives Django a lot of information. With it, Django is able to:

- Create a database schema (CREATE TABLE statements) for this app.
- Create a Python database-access API for accessing ``user`` objects.

Run the following command:

```python
D:\Django_project\HelloWorld> python manage.py makemigrations wkunews
```

You should see something similar to the following:




By running ``makemigrations``, you’re telling Django that you’ve made some changes to your models (in this case, we’ve made a new model) and that you’d like the changes to be stored as a migration.

Migrations are how Django stores changes to your models (and thus your database schema) - they’re files on disk. You can read the migration for your new model if you like; it’s the file ``wkunews/migrations/0001_initial.py`` (the same one shown in your command line). Don’t worry, you’re not expected to read them every time Django makes one, but they’re designed to be human-editable in case you want to manually tweak how Django changes things.

**First, let’s see what SQL that migration would run**. 

The sqlmigrate command takes migration names and returns their SQL:

```python
D:\Django_project\HelloWorld> python manage.py sqlmigrate wkunews "your migration file name"   # "0001_initial" in this case
```

You should see something similar to the following:



**Note**
- Table names are automatically generated by combining the name of the app ("wkunews") and the lowercase name of the model ("user").
- Primary keys (IDs) are added automatically. (You can override this, too)
- The ``sqlmigrate`` command doesn’t actually run the migration on your database - instead, it prints it to the screen so that you can see what SQL Django thinks is required. It’s useful for checking what Django is going to do or if you have database administrators who require SQL scripts for changes.


**Now, run migrate again to create those model tables in your database:** 

```python
D:\Django_project\HelloWorld> python manage.py migrate
```

You should see something similar to the following:



The migrate command takes all the migrations that haven’t been applied (Django tracks which ones are applied using a special table in your database called django_migrations) and runs them against your database - essentially, synchronizing the changes you made to your models with the schema in the database.

Three-step guide to make model changes:
- Change your models (in ``models.py``).
- Run ``python manage.py makemigrations`` to create migrations for those changes.
- Run ``python manage.py migrate`` to apply those changes to the database.

## 4. Django’s own database API

Each view can read records from a database, process and returning an HttpResponse object containing the content for the requested page. Django provides its own database API.

Now, let’s hop into the interactive Python shell and play around with the free API Django gives you. To invoke the Python shell, use this command:

```python
 D:\Django_project\HelloWorld> python manage.py shell
```

Once you’re in the shell, explore the database API: (you can ignore this part if you don't want to access database)

```python
# Import the model class we just wrote.
In [1]: from wkunews.models import user

# objects.all() displays all the users in the database.
# return a QuerySet
In [2]: user.objects.all()
Out[2]: <QuerySet []>

# create a new user
In [3]: u = user(username='testuser', password='123', year=2023)

# save the object into database
In [4]: u.save()

# new object has an id
In [5]: u.id
Out[5]: 1

# Access model field values via Python attributes.
In [8]: u.username
Out[8]: 'testuser'

In [9]: u.password
Out[9]: '123'

In [10]: u.year
Out[10]: 2
    
# A QuerySet is iterable, and it executes its database query the first time you iterate over it.
In [11]: for i in user.objects.all():
   ...:     print(i.username,i.password,i.year)
   ...:
testuser 123 2

In [12]: exit
    
```

Now, modify the login view in ``wkunews/views.py``, in which the view reads information from database user table and check whether the username and password combination is correct or not.

```python
from .models import user

def login(request):
    if request.method == 'POST':
        username = ''
        passwd = ''
        message = ''
        form = LoginForm(request.POST or None)
        if form.is_valid():
            username = form.cleaned_data.get("username")
            passwd = form.cleaned_data.get("password")
            try:
                userinfo = user.objects.get(username=username)
                if userinfo.password == passwd:
                    return HttpResponseRedirect('success')
                else:
                    message = 'Incorrect Password!'
            except:
                message = 'User not exist!'
            return render(request,'login.html',{'form':form, 'message':message}) 
    else:
        form = LoginForm()
    return render(request, 'login.html', {'form':form})
```

In the modified login function, we will search the database for the logged in user. 
- If the user object exists and the password matches, redirect to the success view.
- If the user object does not exist, return the form and fail message.

Also, modify the ``login.html`` template to show the message if login fails:

```python
<h2>Login</h2>

<form method="post">
    {% if message %}
     <div style="color:red" >{{ message }}</div>
    {% endif %}
  
    {% csrf_token %}
    {{ form.as_p }}
  
     <input type="submit">
</form>
```
