## The First Piece of Code

Now we can start to write the first piece of code. 
In Django, every web app you want to create is called a project; and a project is a sum of applications. An application is a set of code files relying on the MVT (Model-View-Template) pattern.

``django-admin`` is Django’s command-line utility for administrative tasks. You can use this command-line utility to create a Django web project. The steps are as follows:

    On your PC (Windows or Mac), open an Anaconda prompt terminal
    Change to a drive you want to save your project, for example: D:  (you may save it in other drive)
    Create a folder called "Django_project": mkdir Django_project
    Change directory to Django_project:  cd Django_project
    Start a new project called HelloWorld on Anaconda Prompt: django-admin startproject HelloWorld

After creating this project, you can find a new folder called "D:\Django_project\HelloWorld" with the following structure:

``HelloWorld/
    HelloWorld/
        __init__.py
        asgi.py  ( Django 3.0 new feature)
        settings.py
        urls.py
        wsgi.py   
    manage.py``

**Folder and file description**:
- ``HelloWorld``: the container for project,it actually contains two elements:
    - ``manage.py``: This file is kind of your project local django-admin for interacting with your project via command line (start the development server, sync db...). To get a full list of command accessible via manage.py you can use the code ``python manage.py help``
    - ``HelloWorld`` subfolder: This folder is the actual python package of your project. It contains four files
        - ``__init__.py``: Just for python, treat this folder as package.
        - ``asgi.py``:   Asynchronous Server Gateway Interface
        - ``settings.py``: As the name indicates, your project settings.
        - ``urls.py``: All links of your project and the function to call.
        - ``wsgi.py``: If you need to deploy your project over WSGI(Python Web Server Gateway Interface).


**Start the development servers**


We need to do migration first. Basically, It will create a default database in sqlite3 for this webserver.
- ``cd HelloWorld``  (cd to the folder where manage.py exists)
- Typing the command: ``python manage.py migrate``

Start Django web server
- Type the command: ``python manage.py runserver``

- open a browser and input the following url: ``127.0.0.1:8000``, if the server starts successfully, you will get the following result.


**Hello world Page**

Create a file called ``views.py`` under the folder "D:\Django_project\HelloWorld\HelloWorld" and type the following code.(you may have different folder)

```python
from django.http import HttpResponse

def hello(request):
    return HttpResponse("Hello world!")
```

This piece of code creates a simple view, in which the function ``hello`` can take an request as the argument and returns an HttpResponse object with "hello world" message.

In the same folder, you can find a file call urls.py, replace the content with following code.

```python
"""HelloWorld URL Configuration

The `urlpatterns` list routes URLs to views. For more information please see:
    https://docs.djangoproject.com/en/3.0/topics/http/urls/
Examples:
Function views
    1. Add an import:  from my_app import views
    2. Add a URL to urlpatterns:  path('', views.home, name='home')
Class-based views
    1. Add an import:  from other_app.views import Home
    2. Add a URL to urlpatterns:  path('', Home.as_view(), name='home')
Including another URLconf
    1. Import the include() function: from django.urls import include, path
    2. Add a URL to urlpatterns:  path('blog/', include('blog.urls'))
"""
from django.contrib import admin
from django.urls import path
from  .  import views

urlpatterns = [
    path('admin/', admin.site.urls),
    path('',views.hello),
]
```

Open your browser, **refresh** your webpage, you should see the hello world message!
