```python
import jieba

#Maybe you want to turn off new word discovery for the phrase:
#string='我们中出了一名叛徒'
#jieba.cut(string, HMM=False)

strings=['南京市长江大桥',
'来到杨过曾经生活的地方，小龙女动情地说：我也想过过过儿过过的生活',
'货拉拉拉不拉拉布拉多，取决于货拉拉拉拉布拉多时拉布拉多拉不拉粑粑',
'校长说：校服上除了校徽别别别的，让你们别别别的别别别的你非别别的',
'明明明明明白白白喜欢他，可她就是不说',
'人要是行，干一行行一行，一行行行行行，行行行干哪行都行。要是不行，干一行不行一行，一行不行行行不行，行行不行干哪行都不行',
'这几天天天天气不好',
'我有一个小本本本来很干净',
'用毒毒毒蛇毒蛇会不会被毒毒死',
'你去班上数数数数数不好的有多少',
'多亏跑了两步，差点没上上上上海的车',
'明明明明明白白白喜欢他，可她就是不说',
'今天下雨，我骑车差点摔倒，好在我一把把把把住了']


#Precise mode
for s in strings:
    text_cut=jieba.cut(s)
    print(" ".join(text_cut))
```

```python
#Full mode
for s in strings:
    text_cut=jieba.cut(s,cut_all=True)
    print(" ".join(text_cut))
```

```python
#Search mode
for s in strings:
    text_cut=jieba.cut_for_search(s)
    print(" ".join(text_cut))
    #type(text_cut)
```

# Remove the stopwords for Chinese

Go through the stopwords dictionary word by word, and remove them

```python
string='总的来说，人说话的时候废话非常多。与此同时，废话多了有害身体健康，换句话说，不如别说废话！'

text_cut = jieba.cut(string)
print(" ".join(text_cut))

def stopwordslist(filepath):  
    stopwords = [line.strip() for line in open(filepath, 'r', encoding='utf-8').readlines()]  
    return stopwords

def seg_sentence(sentence):  
    sentence_seged = jieba.cut(sentence.strip())  
    stopwords = stopwordslist('data\\baidu_stopwords.txt')
    outstr = ''  
    for word in sentence_seged:  
        if word not in stopwords:  
            if word != '\t':  
                outstr += word  
                outstr += " "
    return outstr  

line_seg = seg_sentence(string)
print(line_seg)
type(line_seg)
```

```python
# remove punctuation
# install zhon first:  pip install zhon
from zhon.hanzi import punctuation

for i in punctuation:
    line_seg = line_seg.replace(i, '')

line_seg
```

You can also provide a text file for Jieba Tokenization

```python
textFile = 'data/sanguo.txt'
tokenFile = 'data/sanguo_token.txt'

with open(textFile, 'r', encoding = 'utf-8') as sourceFile, open(tokenFile, 'a+', encoding = 'utf-8') as targetFile:
    for line in sourceFile:
        seg = jieba.cut(line.strip(), cut_all = False)
        output = ' '.join(seg)
        targetFile.write(output)
        targetFile.write('\n')
```

```python
textFile = 'data/santi.txt'
processedFile = 'data/santi_removesw.txt'

inputs = open(textFile, 'r', encoding='utf-8')
outputs = open(processedFile, 'w', encoding='utf-8')
for line in inputs:
    line_seg = seg_sentence(line)
    outputs.write(line_seg + '\n')
```
